### R2 Combined Biomedical (MITDB, SVDB, LTDB)
Purely medical pool — ECG records (arrhythmias, long-term monitoring)

Все датасеты реальные, с point-level labels и разнообразными типами аномалий (point spikes по амплитуде, collective/group, trend, frequency-like/ритм).  

```python
import pandas as pd
import matplotlib.pyplot as plt


def plot_time_series(file_path, start=None, end=None):
    df = pd.read_csv(file_path)
    
    df_slice = df.iloc[start:end]
    
    x = df_slice.index
    y = df_slice['Data']
    labels = df_slice['Label']

    plt.figure(figsize=(12, 6))
    plt.plot(x, y, color='lightblue', linewidth=1.2, zorder=1)
    # plt.scatter(x[labels == 0], y[labels == 0], color='blue', label='Label 0', s=1, zorder=2)
    plt.scatter(x[labels == 1], y[labels == 1], color='red', label='Outliers', s=5, zorder=2)

    plt.title(f'График {file_path} (интервал: {start if start else 'начало'}:{end if end else "конец"})')
    plt.xlabel('Порядковый номер записи')
    plt.ylabel('Значение (Data)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    plt.show()
```

### R1 Combined IT (YAHOO, SMD, IOPS, Exathlon, WSD, NEK)


```python
plot_time_series(f'YAHOO/Yahoo_A1real_34_data.csv')                                    
plot_time_series('SMD/processed_test_machine-2-3_col_2.csv')                            
plot_time_series('IOPS/KPI-1c6d7a26-1f1a-3321-bb4d-7a9d969ec8f0.csv')   

for i in range(3,5):
    try:
        plot_time_series(f'WSD/WSD_{i}.csv')
    except FileNotFoundError: pass

for i in range(1):
    try:
        plot_time_series(f'NEK/NEK_{i}.csv')
    except FileNotFoundError: pass

plot_time_series('Exathlon/2_1_100000_60-20.test_col_0.csv', 12000, 17000)            
plot_time_series('Exathlon/9_3_500000_74.test_col_1.csv') 

```

### R2 Combined Biomedical (MITDB, SVDB, LTDB)


```python
plot_time_series('MITDB/102_col_0.csv', start = 26000, end = 35000)  
plot_time_series('SVDB/872_col_0.csv', 0, 10000)
plot_time_series('LTDB/14046_col_0.csv', 0, 10000)      
```
